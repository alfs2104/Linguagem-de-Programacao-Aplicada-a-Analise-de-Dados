Problema 1: Desenvolvendo uma Calculadora básica

Situação: Você foi selecionado na sua empresa para desenvolver uma calculadora simples que será utilizada por outros colaboradores. Ao iniciar o programa, o sistema deve informar ao usuário que ele precisará fornecer dois valores numéricos para que as operações matemáticas sejam realizadas. Após receber os valores, o programa deverá calcular e exibir os resultados básicos das operações artméticas.

Desenvolva um programa em Python que:

-Exiba uma mensagem explicando o funcionamento da calculadora
-Solicite ao usuário dois valores numéricos

Calcule:
Soma, subtração, multiplicação, divisão, divisão inteira, resto da divisão e potência.

Exiba todos os resultados na tela.

```python

print("======Solicite ao usuário dois valores======")

valor1 = int(input("Digite um valor: "))
valor2 = int(input("Digite outro valor: "))

soma = valor1 + valor2
diferenca = valor1 - valor2 
multiplicacao = valor1 * valor2 
divisao = valor1 / valor2 
divisao_inteira = valor1 // valor2 
resto_divisao = valor1 % valor2 
potencia = valor1 ** valor2

print("A soma é:", soma)
print("A diferença é:", diferenca)
print("A multiplicacao é:", multiplicacao)
print("A divisão é:", divisao)
print("A divisão inteira é:", divisao_inteira)
print("O resto da divisão é:", resto_divisao)
print("A potência é:", potencia)


```
