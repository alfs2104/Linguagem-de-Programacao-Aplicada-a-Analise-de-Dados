Problema 1 — Desempenho de uma Rede de Supermercados

Situação: Uma rede de supermercados registrou o faturamento diário de uma de suas unidades durante cinco dias úteis, ver Tabela. 

A direção estabeleceu que uma unidade apresenta:

• Bom desempenho: média diária igual ou superior a R$ 22.000; 
• Desempenho regular: média entre R$ 19.000 e R$ 21.999,99; 
• Desempenho crítico: média inferior a R$ 19.000. 

Desenvolva um programa que permita analisar os dados da unidade e apresente:
• quantidade de dias analisados; 
• faturamento total; 
• faturamento médio diário; 
• maior faturamento registrado; 
• menor faturamento registrado; 
• classificação do desempenho da unidade conforme os critérios da direção











```python
dia = ["segunda-feira", "terça-feira", "quarta-feira", "quinta-feira", "sexta-feira"]
faturamento = [18500, 21300, 17800, 24600, 27400]


#Quantidade de dias analisados;
len(dia)

# Faturamento Total
sum(faturamento)

# Faturamento Médio Diário;
fat_medio_diario = sum(faturamento) / len(faturamento)

# Maior Faturamento Registrado;
maior_fat = max(faturamento)

# Menor Faturamento Registrado;
menor_fat = min(faturamento)

# Classificação do desempenho da unidade conforme os critérios da direção;
if fat_medio_diario >= 22000:
    print("Desempenho: Bom")
if fat_medio_diario > 19000 and fat_medio_diario <= 21999:
    print("Desempenho: Regular")
else:
    print("Desempenho: Crítico")
    
print("==============================")
    
print("Quantidade de dias analisados: ",len(dia))
print("Faturamento Total: ", sum(faturamento))
print("Faturamento Médio Diário: ", fat_medio_diario)
print("Maior Faturamento: ", maior_fat)
print("Menor Faturamento: ", menor_fat)
```

Problema 2 — Controle de Velocidade

Situação: Um radar registrou a velocidade de um veículo em uma via cujo limite é de 70 km/h.

Desenvolva um programa que:

• Leia a velocidade do veículo; 
• Verifique se ele ultrapassou 70 km/h; 
• Informe se o veículo está dentro do limite ou se o motorista foi multado;
• Calcule a multa considerando R$ 9,00 por km/h excedente;
• Exiba o valor total da multa.
• Calcule o percentual de excesso de velocidade em relação ao limite.




```python
velocidade = float(input("Digite a velocidade do veículo em km/h: "))

limite_velocidade = 70

if velocidade > limite_velocidade:
    multa = velocidade - limite_velocidade
    percentual_multa = ((velocidade - limite_velocidade) / limite_velocidade) * 100
    
    print("Você está acima da velocidade permitida. Multa aplicada!")
    print(f"Valor da multa: R$ {multa * 9:.2f}")
    print(f"Percentual da multa em relação ao limite: {percentual_multa:.2f}%")
else:
    print("Você está dentro do limite de velocidade. Boa viagem!")
    
    

```

Problema 3 — Monitoramento de Produção Industrial

Situação: Uma indústria possui uma linha com capacidade planejada de 500 unidades por turno. Nos 
últimos seis turnos, foram registradas as seguintes produções: 485, 510, 472, 498, 455 e 520 unidades. 
A equipe utiliza a produção média para avaliar o desempenho geral do período, ver Tabela.

Além da classificação geral, existe uma regra de segurança operacional: Se pelo menos um turno 
apresentar produção inferior a 460 unidades, deverá ser emitido um alerta operacional, 
independentemente da classificação obtida pela média.

Desenvolva um programa que apresente:

• armazene as produções em uma lista; 
• determine a quantidade de turnos analisados; 
• calcule a produção total do período; 
• calcule a produção média por turno; 
• identifique a maior e a menor produção registrada; 
• classifique o desempenho geral do período; 
• verifique se deve ser emitido um alerta operacional.




```python
producoes = [485, 510, 472, 498, 455, 520]

capacidade_planejada = 500

qtde_turnos = len(producoes)

producao_media = sum(producoes) / len(producoes)

producao_total_periodo = sum(producoes)

producao_media_turno = producao_total_periodo / qtde_turnos

maior_producao = max(producoes)

menor_producao = min(producoes)


print("Quantidade de turnos analisados: ", qtde_turnos)
print("Produção média por turno: ", producao_media_turno)
print("Produção total no período: ", producao_total_periodo)
print("Maior produção registrada: ", maior_producao)
print("Menor produção registrada: ", menor_producao)

print("==============================")

# Classificação geral do período conforme os critérios da direção;

if producao_media >= 500:
    print("Meta atingida!")
elif producao_media >= 480 and producao_media < 500:
    print("Atenção!")
else:
    print("Abaixo da meta!")


```

```python
import pandas as pd
```

```python

# Montando o dicionário com os dados

dados = {
    "Mês": ["Janeiro", "Fevereiro", "Março", "Abril", "Maio"],
    "Receita": [85000, 92000, 78000, 105000, 88000],
    "Custos_Variaveis": [46000, 51000, 48000, 54000, 52000],
    "Custos_Fixos": [18000, 19000, 20000, 21000, 22000]
}

# Criando o DataFrame a partir do dicionário
df = pd.DataFrame(dados)

# Calculando o Resultado Financeiro: Resultado = Receita - Custos Variáveis - Custos Fixos
df["Resultado"] = df["Receita"] - df["Custos_Variaveis"] - df["Custos_Fixos"]

# Calculando o Indice de Margem Operacional (IMO): (Resultado / Receita) * 100
df["IMO"] = (df["Resultado"] / df["Receita"]) * 100

# Função para classificar o desempenho com base no IMO
def classificar_desempenho(imo):
    if imo >= 25:
        return "Excelente"
    elif 15 <= imo < 25:
        return "Bom"
    elif 8 <= imo < 15:
        return "Regular"
    elif 0 < imo < 8:
        return "Atenção"
    elif imo == 0:
        return "Ponto de equilibrio"
    else:
        return "Prejuízo"


# Criando a coluna de Classificação do Desempenho com base no IMO
df["Classificação"] = df["IMO"].apply(classificar_desempenho)

# Exibindo a tabela 
df
```

```python
# Convertendo para lista e arredondando para 2 casas decimais
# tolist() converte uma coluna do pandas em uma lista padrão no python 

lista_imo = df["IMO"].round(2).tolist()

print(lista_imo)

print("===================================")



# 1. Encontrando os valores máximo e mínimo do IMO
maior = df["IMO"].max()
menor = df["IMO"].min()

print(f"Maior IMO: {maior:.2f}")
print(f"Menor IMO: {menor:.2f}")

# 2. Localizando o mês correspondente (seu código)
# o .loc é o localizador do Pandas. Ele serve para localizar e filtrar linhas e colunas de um dataframe
# usando condiçoes lógicas


mes_melhor_desempenho = df.loc[df["IMO"] == maior, "Mês"].values[0]
mes_pior_desempenho = df.loc[df["IMO"] == menor, "Mês"].values[0]

print(f"Melhor desempenho no mês: {mes_melhor_desempenho}")
print(f"Pior desempenho no mês: {mes_pior_desempenho}")

```
