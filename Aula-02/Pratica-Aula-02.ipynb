#### Estruturas Condicionais

O que é o comando if?

• O comando if permite que o programa execute um bloco de código somente se uma condição for verdadeira 



• Observação: A indentação é obrigatória em Python para que o bloco seja executado
corretamente.

Prática: Vocês são responsáveis por perguntar a temperatura atual ao usuário e se a
temperatura for acima de 30ºC exibir uma mensagem de se hidratar!


```python
temperatura_atual = float(input("Qual é a temperatura atual: "))

if temperatura_atual >= 30:
    print("Cuidado! a temperatura atual é superior à 30°C, recomenda-se hidratar!")
```

##### If e else

Situação: Um sensor mede a temperatura de uma máquina. Se a temperatura for acima de
80ºC, exiba um alerta. Caso contrário, exiba que a máquina está normal.

Nem sempre queremos executar algo apenas quando a condição é verdadeira. Precisamos de
uma ação para casos contrários.



Se a primeira condição for falsa (bloco 1 ), o else (caso contrário) é executado.

📝Prática: Desenvolva um programa que solicite ao usuário um número inteiro e determine se o
número informado é PAR ou ÍMPAR.


```python
num = int(input("Digite um número: "))

if num % 2 == 0:
    print("O número digitado é par.")
else:
    print("O número digitado é ímpar.")
```

📝Prática: O sistema recebe a informação da senha do usuário. Se o usuário digitar
corretamente a senha do sistema, o acesso é permitido. Caso contrário, o acesso deve
ser negado.


```python
senha = int(input("Digite sua senha: "))

if senha == 1234:
    print("Acesso permitido")
else:
    print("Acesso negado! Senha incorreta.")
```

##### Prática com os três comandos (and, or, not)

Situação: Aprovação de uma compra. Uma loja possui um sistema que decide se uma compra 
pode ser aprovada. A compra será aprovada quando:

• o cliente tiver 18 anos ou mais e renda de pelo menos R$ 2.000; 

• ou o cliente for VIP; 

• além disso, o cliente não pode estar bloqueado.


```python
idade = 20
renda = 2500
cliente_vip = False
bloqueado = False

if (idade >= 18 and renda >= 2000 or cliente_vip) and not bloqueado:
    print("Compra aprovada!")
```

📝Prática: Acrescente a lista de produção mais 4 registros, usando .append( ) (manualmente ou 
por input) e verifique novamente a quantidade de registros usando o comando len( ).


```python
producao = [120, 135, 128, 142, 150]

producao.append(111)
producao.append(115)
producao.append(130)
producao.append(125)

quantidade = len(producao)

print(f"{producao}")
print(f"A quantidade de produtos produzidos é: {quantidade}")
```
